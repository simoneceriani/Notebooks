- Consider the following list:

```python
cities = ["San Diego", "Prague", "Cape Town", "Tokyo", "Melbourne"]
print(cities)
```

1. Slice "Prague":

```python
print(cities[1])
```

2. Slice the cities from "Prague" to "Tokyo":

```python
print(cities[1:4])
```

3. Slice "Prague" and "Tokyo":

```python
print(cities[1:4:2])
```

4. Slice the cities from "San Diego" to "Cape Town" (two ways):

```python
print(cities[0:3])
```

```python
print(cities[:3])
```

5. Slice the cities from "Cape Town" to "Melbourne" (two ways):

```python
print(cities[2:5])
```

```python
print(cities[2:])
```

6. Slice "San Diego", "Cape Town", and "Melbourne" (two ways):

```python
print(cities[0:5:2])
```

```python
print(cities[::2])
```

7. Slice "Melbourne" (two ways):

```python
print(cities[4])
```

```python
print(cities[-1])
```

8. Slice the cities from "Prague" to "Tokyo" using negative indices (alternative to example 2):

```python
print(cities[-4:-1])
```

9. Slice the cities from "Tokyo" to "Prague" using positive indices (reverse order):

```python
cities[3:0:-1]
```

10. Slice the cities from "Tokyo" to "Prague" using negative indices (reverse order) (alternative to example 9):

```python
cities[-2:-5:-1]
```

11. Slice all the cities in reverse order:

```python
print(cities[::-1])
```

## Practice

### `Slicing syntax                           What it does`

`list_name[index]                         Extract element in position index`

`list_name[start:stop:step]               Extract from element start (included) to stop (excluded) one each step`

`list_name[:stop:step]                    Extract from element 0 to stop (excluded) one each step`

`list_name[start::step]                   Extract from element start (included) to end one each step`

`list_name[start:stop]                    Extract from element start (included) to stop (excluded)`

`list_name[negative_index]                Extract element abs(negative_index) starting from end`

`list_name[::negative_step]               Extract from end to start in reverse orders, skipping with steps`

`list_name[::-1]                          Extract from end to start in reverse orders`

```python
objects = ["mobile", "t_shirt", "pencil", "laptop", "hat", "ruler", "tv", "pants", "pen"]

# extract clothes
## objects[[1,4,7]] slicing with indices list is not possible

# in this case 
print(objects[1:8:3]) # works

# or with inline syntax
print([objects[i] for i in (1,4,7)]) # with a tuple, preferred because tuple are immutable
print([objects[i] for i in [1,4,7]]) # with a list
print([objects[i] for i in {1,7,4,1,4,7}]) # with a set, note that gets reordered (set order not guaranteed) and remove duplicates
print([objects[i] for i in range(1,8,3)]) # with a range
```
