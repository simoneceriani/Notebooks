- You are at a Chinese restaurant, and at the end of the meal, you get a fortune cookie. There are only three fortune cookies left. Each of them contains a message:

```python
fortune_cookies = ["The man on the top of the mountain did not fall there",
                   "If winter comes, can spring be far behind?",
                   "Land is always on the mind of a flying bird"]
```

- Which fortune cookie will you get? Let the computer decide! To do so, the computer needs a module called random: 

```python
import random
```

- Here is your message when the computer picks a message index:

```python
# pick a message index
message_index = random.randint(0,len(fortune_cookies)-1)
print(message_index)

# get the message
message = fortune_cookies[message_index]
print(message)
```

- And here is your message when the computer directly picks a message:

```python
# pick a message
message = random.choice(fortune_cookies)
print(message)
```

```python
# head / cross toss
coin = ["head", "cross"]
freq = dict(head=0, cross=0)
for _ in range(10000):
  face = random.choice(coin)
  # print(face)
  freq[face] = freq[face] +1

print(freq)
```

```python
# dice toss
freq = {i: 0 for i in range(1, 7)}
for _ in range(10000):
    toss = random.randint(1,6) # extreme included
    freq[toss] = freq[toss] + 1

print(freq)
```

```python
# list of 10 rnd numbers 0-100
lst = [None] * 10
for i in range(len(lst)):
    lst[i] = random.randint(0, 100) # extreme included

print(lst)
```

```python
# create a list of 100 numbers between 5 and 60
lst = [None] * 100
for i in range(len(lst)):
    lst[i] = random.randint(5, 60)

lst_multiple_4 = [x for x in lst if x % 4 == 0]
lst_not_multiple_4 = [x for x in lst if x % 4 != 0]

def map_number_to_index(n):
    return n // 4 - 2 # map 8 --> 0, 12 --> 1, etc 

def map_index_to_number(i):
    return i * 4 + 8

# create count of unique multiple of 4
count_multiple_4_unique = [0 for i in range(8,61,4)]
for m in lst_multiple_4: 
    i_m = map_number_to_index(m)
    count_multiple_4_unique[i_m] = count_multiple_4_unique[i_m] + 1

print(count_multiple_4_unique)

# select the indexes that are non zero and retrieve the mapped number
multiple_4_unique = [map_index_to_number(i) for i, x in enumerate(count_multiple_4_unique) if x != 0]
print(multiple_4_unique)
```
