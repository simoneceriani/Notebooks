## 1. Arithmetic operations on list elements

- Sum two lists element‐wise:


```python
odd_numbers  = [1,3,5]
even_numbers = [2,4,6]
summed  = []

for i in range(len(odd_numbers)):
    summed.append(odd_numbers[i] + even_numbers[i])

print(summed)
```

- Sum a number to each element of a list:

```python
numbers = [1,2,3] 
number = 3

for i in range(len(numbers)):
    numbers[i] += number 
    
print(numbers)
```

## 2. "Arithmetic" operations between lists

- Concatenate two lists:

```python
odd_numbers  = [1,3,5] # two lists
even_numbers = [2,4,6]
concatenated = odd_numbers + even_numbers
print(concatenated)
```

- Replicate a list 3 times:

```python
numbers = [1,2,3]
number  = 3
replicated = numbers * number
print(replicated)
```

- When is replication useful?

```python
short_list = [0]
number     = 50
long_list  = short_list * number
print(long_list)
```

## 3. List assignment

- Given a list containing a few integers:  

```python
given_list = [1,2,3]
print(given_list)
```

## 4. Adding an element or a list to a list  


- Add one element at the end of a list:

```python
numbers = [1,2,3]
numbers.append(4)
print(numbers)
```

- Insert the number 2 in position 1:

```python
numbers = [1,3,4]
numbers.insert(1,2)
print(numbers)
```

- Concatenate two lists:

```python
first_list  = [1,2,3]
second_list = [4,5,6]
third_list  =  first_list + second_list
print(third_list)
```

- Add one list at the end of another list:

```python
first_list  = [1,2,3]
second_list = [4,5,6]
first_list.extend(second_list)
print(first_list)
```

## 5. Removing elements from a list  

- From the following list, remove all the elements `"ciao"`:

```python
greetings = ["ciao", "ciao", "hello"] 
greetings.remove("ciao")
print(greetings)
```

- How do we remove both `"ciao"` from greetings?

```python
greetings = ["ciao", "ciao", "hello"] 
while "ciao" in greetings:
    greetings.remove("ciao")
print(greetings)
```

- Remove the string `"hello"` based on its position:

```python
greetings = ["ciao", "ciao", "hello"] 
greetings.pop(2)
print(greetings)
```

- Remove all elements in a list:

```python
greetings = ["ciao", "ciao", "hello"] 
greetings.clear()
print(greetings) 
```

## 6. Sorting a list  

- Sort the following list of integers:

```python
numbers = [5,7,6]
numbers.sort()
print(numbers)
```

- Sort the following list of integers in a descending way:

```python
numbers = [5,7,6]
numbers.sort(reverse=True)
print(numbers)
```

- Reverse the following list of integers:

```python
numbers = [5,7,6] 
numbers.reverse()
print(numbers)
```

## 7. Searching and counting elements 

- Create a list and search for a specific element:

```python
letters = ["a","g","c","g"]
position = letters.index("g") 
print(position)
```

- How do we find all positions?

```python
letters = ["a","g","c","g"]
positions = []
for i in range(len(letters)):
    if letters [i] == "g":
        positions.append(i)
print(positions)
```

- Count how many times an element is present in a list: 

```python
letters = ["a","g","c","g"]
n = letters.count("g") 
print(n)
```

```python
items = ["carrots", "zucchini", "potatoes"]
nums = [10,12,11]
prices = [.7, .5, .2]

initial_nums = nums.copy()

total_per_customers = [0] * 3

def printState():
    for [i,n,p] in zip(items,nums,prices):
        print(f"{i}\t{n}\t{p}\t")

def buy(item : str, num : int) -> float:
    if(num == 0): 
        return 0.
            
    if item not in items:
        print(f'item {item} does not exists')
        return 0.0
    idx = items.index(item)
    if(nums[idx] == 0):
        print(f'item {item} is out of stock')
        return 0.0
    elif(nums[idx] < num):
        print(f'item {item} max quantity is {nums[idx]}')
        num = nums[idx]
    nums[idx] = nums[idx] - num
    return num * prices[idx]

printState()

customer_id = 0
buy_items = items.copy()
buy_nums = [2,4,3]
for [i,n] in zip(buy_items, buy_nums):
    total_per_customers[customer_id] = total_per_customers[customer_id] + buy(i, n)

print(total_per_customers)
printState()

customer_id = 1
buy_items = items.copy()
buy_nums = [0,3,3]
for [i,n] in zip(buy_items, buy_nums):
    total_per_customers[customer_id] = total_per_customers[customer_id] + buy(i, n)

print(total_per_customers)
printState()

customer_id = 2
buy_items = items.copy()
buy_nums = [6,4,1]
for [i,n] in zip(buy_items, buy_nums):
    total_per_customers[customer_id] = total_per_customers[customer_id] + buy(i, n)

print(total_per_customers)
printState()

# avg
avg = 0
for p in total_per_customers:
    avg = avg + p
avg = avg / len(total_per_customers)
print(f"avg spent {avg}")

# how many sold
sold = [i - n for [i,n] in zip(initial_nums, nums)]
print(f"sold items {sold}")
# find max idx
idx = 0
for i in range(1,len(sold)):
    if(sold[i] > sold[idx]):
        idx = i
print(f"most sold is {items[idx]} with {sold[idx]}")
# find min idx
idx = 0
for i in range(1,len(sold)):
    if(sold[i] < sold[idx]):
        idx = i
print(f"least sold is {items[idx]} with {sold[idx]}")
```

```python
# reverse 
numbers = [x for x in range(0,10)]
print(numbers)

n_r = numbers.copy()
n_r.reverse()
print(n_r)

n_r = numbers[-1::-1]
print(n_r)

# or
n_r = numbers[::-1]
print(n_r)

# incorrect list comphrension
# n_r = [x for i in range(len(numbers)) x = numbers[len(numbers) - 1 - i]]

n_r = [numbers[len(numbers) - 1 - i] for i in range(len(numbers))]
print(n_r)

# alternative
n_r = list(reversed(numbers))
print(n_r)
```
