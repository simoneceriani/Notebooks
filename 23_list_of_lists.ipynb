## 2. Browsing with nested for loops

- Given the following list of lists: 

```python
sports = [["skiing", "skating", "curling"], ["canoeing", "cycling", "swimming", "surfing"]]
```

- Print the sub-list elements one-by-one using a nested for loops through *indices*:

```python
for i in range(len(sports)):  
    for j in range(len(sports[i])):
        print(sports[i][j])
```

- Print the sub-list elements one-by-one using a nested for loops through *values*:

```python
for seasonal_sports in sports:
    for sport in seasonal_sports:
        print(sport)
```

## 3. Flattening

- Given the following list of lists:

```python
instruments = [["contrabass", "cello", "clarinet"],["gong", "guitar"],["tambourine", "trumpet", "trombone", "triangle"]]
```

- Flatten the list using a nested for loop through *indices*:

```python
flat_instruments = []
for i in range(len(instruments)):
    for j in range(len(instruments[i])):
        flat_instruments.append(instruments[i][j])
print(flat_instruments)
```

- Flatten the list using a nested for loop through  *elements*:

```python
flat_instruments = []
for group in instruments:
    for instrument in group:
        flat_instruments.append(instrument)
print(flat_instruments)
```

- Flatten the list using a for loop and list concatenation:

```python
flat_instruments = []
for group in instruments:
    flat_instruments += group
print(flat_instruments)
```

- Flatten the list using list comprehension:

```python
instruments = [instrument for group in instruments for instrument in group]
print(instruments)
```

```python
numbers = [[3,7,1],[7,6,5,4],[8,9,7,4,5]]

# a. How long is each sub-list?
ns = [len(l) for l in numbers]
print(ns)

#b. In the first sub-list, replace the third element with the sum of the previous two elements.
numbers[0][2] = numbers[0][0] + numbers[0][0]
print(numbers)

#c. In the second sub-list, sort the elements in ascending order.
numbers[1].sort()
print(numbers)

#d. In the third sub-list, substitute the number 4 with the number 3.
bck = numbers[2].copy()
def subs4with3(n):
    if n == 4:
        return 3
    return n
numbers[2] = [subs4with3(n) for n in numbers[2]]
print(numbers)

# or
numbers[2] = bck.copy()
# print(numbers)
numbers[2] = [3 if n == 4 else n for n in numbers[2]]
print(numbers)


#e. How many number 7 are there in total? Save their positions in a list of lists (expected result: [[0, 1],[1, 3], [2, 2]]).
c = []
for i,l in enumerate(numbers):
    c7 = l.index(7)
    c.append([i,c7])
print(c)

# or inline
c = [[i,l.index(7)] for i,l in enumerate(numbers)]
print(c)
```

```python
# Summing up. Given the following list of lists:
numbers = [[1,3,5],[7,2,8],[3,4,9]]

#a. Create a list containing the sum of the numbers in each sub-list (expected result: [9, 17, 16]).
ss = []
for l in numbers:
    s = 0
    for c in l:
        s = s + c
    ss.append(s)
print(ss)

ss = [sum(l) for l in numbers]
print(ss)

#b. Sum all the elements of the list of lists using (1) a for loop through indices and (2) a for loop through elements.
s = 0
for i in range(len(numbers)):
    for j in range(len(numbers[i])):
        s = s + numbers[i][j]
print(s)

s = 0
for l in numbers:
    for e in l:
        s = s + e
print(s)

s = sum([sum(l) for l in numbers])
print(s)
```

```python
# Matrix time! Give the following matrix:
matrix = [[4,1,3,9], [2,1,6,5], [4,0,3,8], [7,2,6,2]]

# a. Print the matrix as a 4x4 table (expected result:
for row in matrix:
    for el in row:
        print(el, end = " ")
    print("")
    
#b. Multiply all the elements on the main diagonal and print the result (expected result: 24). Note: The main
#diagonal goes from top-left to bottom-right. In this example, the main diagonal contains: 4,1,3,2.
d = 1
for i in range(len(matrix)):
    d = d * matrix[i][i]
print(d)

# or
import math
d = math.prod(matrix[i][i] for i in range(len(matrix)))
print(d)

#c. Sum the matrix values vertically (expected result: [17, 4, 18, 24]).
cols = [0] * len(matrix[0])
for row in matrix:
    print(row)
    for j in range(len(row)):
        cols[j] = cols[j] + row[j]
print(cols)


# or
cols = [sum(col) for col in zip(*matrix)]
# explanation:
# * unpack a list and convert to multiple args, it is equivalent to call zip(matrix[0], matrix[1], matrix[2], matrix[3])
# zip return a list element per element of all arguments (all 0 elements, all 1 elements, etc...)
# therefore it is equivalent to a traspose and the for loop iterates on columns
# then sum do the trick
print(cols)
```
